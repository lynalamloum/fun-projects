```python
import random

def mot_aleatoire() :
    
    mots = [
    "python", "ordinateur", "algorithme", "clavier", "variable",
    "fonction", "boucle", "condition", "terminal", "bibliotheque",
    "compilation", "debugger", "interface", "heritage", "recursion",
    "pointeur", "memoire", "processeur", "reseau", "architecture"
    ]
    
    return random.choice(mots)

def mot_masqué(mot) :
    
    mot_guess = list(mot)
    mot_masqué = []
    
    for i in range (len(mot_guess)) :
        mot_masqué.append("_")
    
    return mot_masqué
        
def verif_lettre(lettre, mot) :
    
    for i in range (len(mot)) :
        if mot[i] == lettre :
            return True
    return False

def lettres_trouvées(mot) :
    trouvées = 0
    for i in range (len(mot)) :
        if mot[i] != "_" :
            trouvées += 1
    return trouvées

def maj_lettre(mot_hide, mot, lettre) :
    
    nv_mot = []
    
    for i in range (len(mot)) :
        if mot[i] == lettre :
            nv_mot.append(lettre)
        else :
            nv_mot.append(mot_hide[i])
            
    return nv_mot

def afficher_mot(nv_mot) :
    
    print(" ".join(nv_mot)) 
            
def bonhomme(erreur) :
    
    if (erreur == 0) :
        
        print(" ")
    
    elif(erreur == 1) :
        
        print(r"""     |""")
        
    elif(erreur == 2) :
        
        print(r"""     |
     O""")
        
    elif(erreur == 3) :
        
        print(r"""     |
     O
     |""")
        
    elif(erreur == 4) :
        
        print(r"""     |
     O
     |
    \ /""")
        
    elif(erreur == 5) :
        
        print(r"""     |
     O
     |
    \ /
     |""")
        
    else :
        
        print(r"""     |
     O
     |
    \ /
     |
    / \ """)
        
    
    
def pendu() :
    
    mot = mot_aleatoire()
    mot_hide = mot_masqué(mot)
    tentatives = 1
    win = False
    erreurs = 0
    trouvées = 0
    
    print(f"{tentatives}e tentative")
    
    print(r"""     |
     O
     |
    \ /
     |
    / \ """)
    
    afficher_mot(mot_hide)
    
    while(tentatives <= 6 and not win) :
        
        print(" ")
        lettre = str(input("Choisissez une lettre : "))
        
        if verif_lettre(lettre, mot) == True :
            mot_hide = maj_lettre(mot_hide, mot, lettre)
            trouvées = lettres_trouvées(mot_hide)
            
        else :
            tentatives += 1
            erreurs += 1
    
        if "_" not in mot_hide:
            win = True

        if not win:
            print(f"{tentatives}e tentative")
            print(" ")
            bonhomme(erreurs)
            print(" ")
            afficher_mot(mot_hide)
        
    print(" ")
    print(mot)
    print(" ")
    
    if win :
        print("Gagné!")
    else :
        print("Perdu..")
        
pendu()
```
